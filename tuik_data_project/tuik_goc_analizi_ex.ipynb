```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
import pandas as pd
import plotly.express as px

df = pd.read_excel("Göç etme nedenine göre illerin aldığı göç.xls", skiprows=4, header=None)
df.columns = [
    "İl", "Toplam", "Tayin/iş değişikliği", "İş bulmak/işe başlamak",
    "Eğitim", "Evlilik", "Ailevi nedenler", "Daha iyi konut/yaşam",
    "Sağlık", "Hanedeki fertlere bağımlı", "Diğer", "Bilinmeyen",
    "Sütun12", "Sütun13"
]

turkey = df[df["İl"] == "Toplam-Total"].iloc[0]
turkey_data = turkey[2:].drop(["Bilinmeyen", "Sütun12", "Sütun13"])
turkey_data = turkey_data.sort_values(ascending=True)

fig = px.bar(
    x=turkey_data.values,
    y=turkey_data.index,
    orientation="h",
    title="Reasons for Migration to Turkey (2025)",
    labels={"x": "Number of People", "y": "Reason"},
    color=turkey_data.values,
    color_continuous_scale="Viridis"
)
fig.show()
```

```python
import plotly.graph_objects as go

istanbul = df[df["İl"] == "İstanbul"].iloc[0][2:]
ankara = df[df["İl"] == "Ankara"].iloc[0][2:]

drop_cols = ["Bilinmeyen", "Sütun12", "Sütun13"]
istanbul = istanbul.drop(drop_cols)
ankara = ankara.drop(drop_cols)

fig = go.Figure()
fig.add_trace(go.Bar(
    y=istanbul.index,
    x=istanbul.values,
    name="İstanbul",
    orientation="h",
    marker_color="steelblue"
))
fig.add_trace(go.Bar(
    y=ankara.index,
    x=ankara.values,
    name="Ankara",
    orientation="h",
    marker_color="coral"
))
fig.update_layout(
    title="Reasons for Migration: Istanbul vs Ankara (2025)",
    xaxis_title="Number of People",
    barmode="group",
    height=600
)
fig.show()
```

```python
top10 = df.nlargest(11, "Toplam")[["İl", "Toplam"]].iloc[1:]

fig = px.bar(
    top10,
    x="Toplam",
    y="İl",
    orientation="h",
    title="Top 10 Cities by In-Migration (2025)",
    labels={"Toplam": "Number of People", "İl": "City"},
    color="Toplam",
    color_continuous_scale="Blues"
)
fig.show()
```

```python
df_veren = pd.read_excel("Göç etme nedenine göre illerin verdiği göç.xls", skiprows=4, header=None)
df_veren.columns = df.columns

df["Net_Göç"] = df["Toplam"] - df_veren["Toplam"]
df_iller = df[df["İl"] != "Toplam-Total"].copy()

# En çok net göç alan ve veren 10 il
top_net_alan = df_iller.nlargest(10, "Net_Göç")[["İl", "Net_Göç"]]
top_net_veren = df_iller.nsmallest(10, "Net_Göç")[["İl", "Net_Göç"]]

fig = go.Figure()
fig.add_trace(go.Bar(
    y=top_net_alan["İl"],
    x=top_net_alan["Net_Göç"],
    name="Net Göç Alan",
    orientation="h",
    marker_color="green"
))
fig.add_trace(go.Bar(
    y=top_net_veren["İl"],
    x=top_net_veren["Net_Göç"],
    name="Net Göç Veren",
    orientation="h",
    marker_color="red"
))
fig.update_layout(
    title="Net Migration by City (2025)",
    xaxis_title="Net Migration",
    height=600,
    barmode="group"
)
fig.show()
```
